```python
#use UPanda internal emualtor model
from UPanda import integrated_model_jax as integrated_model

tt_emu_model = integrated_model.IntegratedModel(None, None, None, None)
tt_emu_model.restore("../example_models/integrated_models/tt_emu")

ee_emu_model = integrated_model.IntegratedModel(None, None, None, None)
ee_emu_model.restore("../example_models/integrated_models/ee_emu")

te_emu_model = integrated_model.IntegratedModel(None, None, None, None)
te_emu_model.restore("../example_models/integrated_models/te_emu")
```

```python
additional_args = {'add_lowl_tt':False, 'integrated_model':True, "add_lowl_tt":True}
```

```python
total_emu_model_list = [[tt_emu_model, te_emu_model, ee_emu_model], [tt_emu_model, te_emu_model, ee_emu_model]]
```

```python
from UPanda.module import * 
```

```python
params_chain, loglike =  UPanda.run_panda.run_mcmc(
    total_emu_model_list = total_emu_model_list,
    experiments_list = ["hillipop", "lollipop"],
    compression_list = [[], []],
    filename_config="../configs/config_example_cmb_planck_pr4.yaml",
    n_samples = 2e6,
    n_walkers=64,
    sampler = "emcee", 
    additional_args=additional_args)
```

```python
tri = TriangleChain(density_estimation_method='smoothing', label_fontsize=50)
tri.contour_cl(params_chain, color='steelblue', label = r"planck CMB primary + CMB lensing", label_fontsize=28, show_values=True)
```
